## **Existencia de Tablas**

```python
class TestTables(DatabaseTestCase):
    """Prueba la existencia de las tablas en la base de datos."""
    def test_tables_exist(self):
        self.cursor.execute("SHOW TABLES")
        tables = {table[0] for table in self.cursor.fetchall()}
        expected_tables = {
            "conference", "team", "player", "stadium", "fault", "season",
            "game", "game_team", "player_fault", "team_position", "player_point"
        }
        self.assertEqual(tables, expected_tables, "Las tablas existentes no coinciden con las esperadas")
```

## **Consultas Específicas**

```python
class TestQueries(DatabaseTestCase):
    """Verifica que las consultas SQL devuelvan los resultados esperados."""
    def test_players_and_teams(self):
        """Obtiene la lista de jugadores junto con sus equipos."""
        query = """
            SELECT p.id, p.name, t.name
            FROM player p
            INNER JOIN team t ON p.team_id = t.id;
        """
        self.cursor.execute(query)
        result = self.cursor.fetchall()
        self.assertGreaterEqual(len(result), 1, "La consulta de jugadores y equipos no devolvió resultados")

    def test_games_and_stadiums(self):
        """Obtiene la lista de partidos junto con los estadios donde se jugaron."""
        query = """
            SELECT g.id, g.date, s.name
            FROM game g
            INNER JOIN stadium s ON g.stadium_id = s.id;
        """
        self.cursor.execute(query)
        result = self.cursor.fetchall()
        self.assertGreaterEqual(len(result), 1, "La consulta de partidos y estadios no devolvió resultados")
```

```python
print("\n=== EJECUCIÓN: TestQueries ===\n")
suite_queries = unittest.TestLoader().loadTestsFromTestCase(TestQueries)
result_queries = unittest.TextTestRunner(verbosity=2).run(suite_queries)
```

## **Validaciones Lógicas**

```python
class TestBooleanAssertions(DatabaseTestCase):
    """Ejemplos de uso de assertTrue y assertFalse en consultas a la BD."""

    def test_no_team_without_conference(self):
        """Verifica que no existan equipos sin conferencia asignada (conference_id)."""
        query = "SELECT id FROM team WHERE conference_id IS NULL"
        self.cursor.execute(query)
        rows = self.cursor.fetchall()
        self.assertTrue(len(rows) == 0, "Se encontraron equipos sin conferencia asignada")

    def test_at_least_one_game_in_season(self):
        """Comprueba que haya al menos un juego en la temporada con id=1."""
        query = "SELECT id FROM game WHERE season_id = 1"
        self.cursor.execute(query)
        rows = self.cursor.fetchall()
        self.assertFalse(len(rows) == 0, "No se encontraron juegos para season_id=1")

    def test_player_ids_positive(self):
        """Verifica que todos los jugadores tengan un ID mayor a 0 (ninguno debe ser <= 0)."""
        query = "SELECT id FROM player"
        self.cursor.execute(query)
        rows = self.cursor.fetchall()
        all_positive = all(r[0] > 0 for r in rows)
        self.assertTrue(all_positive, "Se encontró al menos un jugador con ID <= 0")
```

```python
print("\n=== EJECUCIÓN: TestBooleanAssertions ===\n")
suite_boolean = unittest.TestLoader().loadTestsFromTestCase(TestBooleanAssertions)
result_boolean = unittest.TextTestRunner(verbosity=2).run(suite_boolean)
```

## **Integridad Referencial**


```python
class TestReferentialIntegrity(DatabaseTestCase):
    """Comprueba que no existan registros huérfanos en game_team."""

    def test_no_orphan_game_team(self):
        """
        Verifica que la tabla game_team no tenga 'game_id' o 'team_id'
        apuntando a registros inexistentes en las tablas 'game' o 'team'.
        """
        query = """
            SELECT gt.id
            FROM game_team gt
            LEFT JOIN game g ON gt.game_id = g.id
            LEFT JOIN team t ON gt.team_id = t.id
            WHERE g.id IS NULL
               OR t.id IS NULL
        """
        self.cursor.execute(query)
        orphans = self.cursor.fetchall()
        self.assertTrue(
            len(orphans) == 0,
            f"Se encontraron filas huérfanas en game_team: {orphans}"
        )
```

```python
print("\n=== EJECUCIÓN: TestReferentialIntegrity ===\n")
suite_ref_int = unittest.TestLoader().loadTestsFromTestCase(TestReferentialIntegrity)
result_ref_int = unittest.TextTestRunner(verbosity=2).run(suite_ref_int)
```
